## Connecting the control computer to the xArm controller

Before running the driver, confirm that the computer is on the same network as the xArm controller.

1. Assign the network interface connected to the robot an IP address on the same subnet. 
 > For example, on macOS: `sudo ifconfig en8 inet 192.168.1.10 netmask 255.255.255.0`
2. Optionally reset the interface to DHCP:
   `sudo ipconfig set en11 DHCP`
3. Test connectivity:
   `ping 192.168.1.201`
4. In the UFactory app, verify that the robot is connected and set to `real robot`.
5. Replace `en11` with the correct interface name for your machine.

> Note:
> - The exact interface name depends on your OS and Ethernet adapter.
> - You may be prompted for administrator authentication when changing network settings.
